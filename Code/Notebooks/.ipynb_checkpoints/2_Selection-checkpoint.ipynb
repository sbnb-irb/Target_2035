## This notebook exemplifies how to **select** the **two pocket representatives** from a given cluster according to the two presented selection strategies. 

```python
outpath = Path("Examples/2_Selection/")
cluster = 1783
cluster_data = pd.read_csv(outpath / ("cluster_" + str(cluster) + ".tsv"), sep="\t")
```

### 1. Ligand-associated data

First, we have to load the different metrics of interest, separate the pockets according to their ligand-assocaition status and compute the buriedness threshold. 

This step is necessary for both strategies.

```python
pocket_metrics = pd.read_csv("Results/Selection/AF2_PD_ligand_information.tsv", sep="\t")
lig_associated_metrics = pocket_metrics[pocket_metrics["Ligand-associated"] == True]
```

```python
buriedness_threshold = compute_buriedness_threshold(lig_associated_metrics["Buriedness"])
print(buriedness_threshold)
```

### 2. Novelty selection

```python
# Apply the appropiate filters
filtered_pockets = apply_novelty_filter(cluster_data, pocket_metrics, buriedness_threshold)
```

```python
print(len(filtered_pockets))
```

```python
# Compute the mahalanobis distance of the pockets
compute_mahalanobis(filtered_pockets, lig_associated_metrics)
```

```python
# Select the cluster representatives
outfile = outpath / ("cluster_" + str(cluster) + "_novel_selection.tsv")
choose_novel_representatives(filtered_pockets, outfile)
```

### 3. Experimental selection

```python
# Apply the appropiate filters
filtered_pockets = apply_experimental_filter(cluster_data, pocket_metrics, buriedness_threshold)
```

```python
print(len(filtered_pockets))
```

```python
# Compute the mahalanobis distance of the pockets
compute_mahalanobis(filtered_pockets, lig_associated_metrics)
```

```python
# Select the cluster representatives
outfile = outpath / ("cluster_" + str(cluster) + "_experimental_selection.tsv")
choose_experimental_representatives(filtered_pockets, outfile)
```
