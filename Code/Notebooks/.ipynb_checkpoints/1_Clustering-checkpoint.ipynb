## This notebook exemplifies how to define **Pfam-level pockets**, obtain the **pocket representative**, identify if the pockets are **associated to ligands** and finally **cluster** the pocket representatives.

Prior to anything, you have to obtain the structures, define its centroids (either through a know ligand or *de novo*), prepare the structures and generate the PocketVec descriptors as detailed in the original PocketVec publication ([Github](https://github.com/sbnb-irb/PocketVec/tree/master)):

Comajuncosa-Creus, A., Jorba, G., Barril, X., & Aloy, P. (2024). Comprehensive detection and characterization of human druggable pockets through binding site descriptors. Nature Communications, 15(1), 7917.

```python
outpath = "Examples/1_Clustering/"
pfam = "PF00006"
```

### 1. Define Pfam-level pockets

This step requires to:

1. Convert the centroids from SD to PDB.
2. Randomly select a reference structure.
3. Superimpose the structures on top of the chosen reference.
4. Cluster the superimposed centroids to obtain Pfam-level pockets. 

```python
structure_path = Path(outpath) / pfam
```

```python
domains = sorted(p.name for p in structure_path.iterdir()if (p / f"{p.name}.pdb").exists())
```

```python
# Centroid conversion 
convert_pockets(structure_path, domains)
```

```python
# Select reference structure
reference = select_reference(domains)
print(reference)
```

```python
# Structure superposition
superpose_structures(structure_path, domains, reference)
```

```python
# Cluster the centroids
cluster_centroids(structure_path)
```

### 2. Choose a representative

```python
structure_path = Path(outpath) / pfam
```

```python
# Load the generated Pfam-level pockets and the descriptors
pfam_pockets = pd.read_csv(structure_path / (pfam + "_pockets.tsv"), sep = "\t")
descriptors = pickle.load(open("Results/Descriptors/descriptors_AF2_PD_80.pkl", "rb"))
```

```python
# For each Pfam pocket obtain its representative
representative_data = [] 
for i, data in pfam_pockets.iterrows():
    included_pockets = data["Included pockets"].split(",")
    output = find_representative(included_pockets, descriptors)
    representative_data.append(output)

pfam_pockets[["Representative", "Mean CD", "Representative mean CD"]] = representative_data
```

```python
# Sve the dataset
pfam_pockets.to_csv(structure_path / (pfam + "_representative_pockets.tsv"), sep = "\t", index = False)
```

### 3. Identifying ligand-associated pockets

This step requires to:

1. Convert the centroids from SD to PDB.
2. Cluster the LIG and PD centroids to obtain the ligand-associated pockets.


```python
structure_path = Path(outpath) / (pfam + "_Lig")
```

```python
domains = sorted(p.name for p in structure_path.iterdir()if (p / f"{p.name}.pdb").exists())
```

```python
# Centroid conversion 
convert_pockets(structure_path, domains)
```

```python
# Cluster the centroids
for domain in domains:
    cluster_centroids_lig(structure_path / domain)
```

### 4. Clustering the reduced dataset

```python
pfam_pockets = pd.read_csv("Results/Selection/representative_pockets.tsv", sep="\t")
representative_pockets = pfam_pockets["Representative"].to_list()
descriptors = pickle.load(open("Results/Descriptors/descriptors_AF2_PD_80.pkl", "rb"))
```

```python
outfile = Path(outpath) / "clustered_reduced_descriptors.csv"
```

```python
cluster_pockets(representative_pockets, descriptors, outfile)
```
